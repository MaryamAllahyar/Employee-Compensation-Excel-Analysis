# Employee Compensation & Pay Adjustment Analysis

## Data Generation

This notebook creates a synthetic employee dataset for an Excel-based compensation analysis project.

The dataset is fictional and is designed to simulate workforce compensation data while avoiding the use of confidential employee information.

```python
import pandas as pd
import numpy as np

np.random.seed(42)
```

```python
job_data = {
    "Administration": [
        ("Administrative Assistant", "L1"),
        ("Administrative Coordinator", "L2"),
        ("Office Manager", "L3")
    ],
    
    "Finance": [
        ("Finance Assistant", "L1"),
        ("Financial Analyst", "L2"),
        ("Senior Financial Analyst", "L3")
    ],
    
    "Human Resources": [
        ("HR Assistant", "L1"),
        ("HR Analyst", "L2"),
        ("HR Advisor", "L3")
    ],
    
    "Information Technology": [
        ("IT Support Analyst", "L1"),
        ("Systems Analyst", "L2"),
        ("Senior Systems Analyst", "L3")
    ],
    
    "Policy & Research": [
        ("Research Assistant", "L1"),
        ("Policy Analyst", "L2"),
        ("Senior Policy Analyst", "L3")
    ],
    
    "Operations": [
        ("Operations Assistant", "L1"),
        ("Operations Analyst", "L2"),
        ("Operations Manager", "L3")
    ]
}
```

```python
#create a fictional external market benchmarks
market_salaries = {
    "Administrative Assistant": 48000,
    "Administrative Coordinator": 58000,
    "Office Manager": 72000,

    "Finance Assistant": 52000,
    "Financial Analyst": 70000,
    "Senior Financial Analyst": 88000,

    "HR Assistant": 50000,
    "HR Analyst": 68000,
    "HR Advisor": 82000,

    "IT Support Analyst": 58000,
    "Systems Analyst": 78000,
    "Senior Systems Analyst": 98000,

    "Research Assistant": 52000,
    "Policy Analyst": 72000,
    "Senior Policy Analyst": 90000,

    "Operations Assistant": 50000,
    "Operations Analyst": 68000,
    "Operations Manager": 85000
}
```

```python
#Generate 500 employees
employees = []

for i in range(1, 501):

    department = np.random.choice(list(job_data.keys()))
    job_title, job_level = job_data[department][
        np.random.randint(len(job_data[department]))
    ]

    years_service = np.random.randint(0, 21)

    performance_rating = np.random.choice(
        [1, 2, 3, 4, 5],
        p=[0.05, 0.10, 0.50, 0.25, 0.10]
    )

    market_median = market_salaries[job_title]

    # Generate salary around the market median
    market_position = np.random.normal(0.95, 0.10)

    current_salary = round(
        market_median * market_position,
        -2
    )

    employees.append([
        f"E{i:04d}",
        department,
        job_title,
        job_level,
        "Halifax",
        years_service,
        current_salary,
        performance_rating
    ])

employee_df = pd.DataFrame(
    employees,
    columns=[
        "Employee_ID",
        "Department",
        "Job_Title",
        "Job_Level",
        "Location",
        "Years_Service",
        "Current_Salary",
        "Performance_Rating"
    ]
)

employee_df.head()
```
